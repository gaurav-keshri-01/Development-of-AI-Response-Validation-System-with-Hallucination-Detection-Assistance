1. Dictionary: Safely Access a Missing Key

**Question:** What is the difference between dictionary indexing and `get()` when a key is missing?

**Python code:**

```python
student = {
    "name": "Rahul",
    "age": 21,
    "branch": "CSE"
}

print(student["name"])
print(student.get("marks"))
print(student.get("marks", 0))
```

**Expected output:**

```text
Rahul
None
0
```

**Explanation:** `student["name"]` returns the value for an existing key. `get("marks")` returns `None` when the key is missing, while `get("marks", 0)` returns the supplied default value. Using `student["marks"]` would raise a `KeyError`.

````python
student = {
    "name": "Rahul",
    "age": 21,
    "branch": "CSE"
}

print(student["name"])
print(student.get("marks"))
print(student.get("marks", 0))
````

2. Dictionary: Count Element Frequencies

**Question:** Count how many times each number appears in a list.

**Python code:**

```python
numbers = [1, 2, 2, 3, 1, 2, 4]
frequency = {}

for number in numbers:
    frequency[number] = frequency.get(number, 0) + 1

print(frequency)
```

**Expected output:**

```text
{1: 2, 2: 3, 3: 1, 4: 1}
```

**Explanation:** Each number becomes a dictionary key and its count becomes the value. `get(number, 0)` starts the count at zero for a number that has not appeared before. This approach takes O(n) average time for n input elements.

````python
numbers = [1, 2, 2, 3, 1, 2, 4]
frequency = {}

for number in numbers:
    frequency[number] = frequency.get(number, 0) + 1

print(frequency)
````

3. Tuple: Immutability

**Question:** What happens when you try to assign a new value to a tuple element?

**Python code:**

```python
values = (10, 20, 30)

try:
    values[0] = 100
except TypeError as error:
    print(type(error).__name__)
    print(error)

print(values)
```

**Expected output:**

```text
TypeError
'tuple' object does not support item assignment
(10, 20, 30)
```

**Explanation:** Tuples do not allow their elements to be reassigned. The `try`/`except` block catches the expected `TypeError`, so the notebook continues executing instead of stopping at this example.

````python
values = (10, 20, 30)

try:
    values[0] = 100
except TypeError as error:
    print(type(error).__name__)
    print(error)

print(values)
````

4. Tuple: Single-Element Tuple

**Question:** What is the difference between `(5)` and `(5,)`?

**Python code:**

```python
a = (5)
b = (5,)

print(type(a).__name__)
print(type(b).__name__)
print(b)
```

**Expected output:**

```text
int
tuple
(5,)
```

````python
a = (5)
b = (5,)

print(type(a).__name__)
print(type(b).__name__)
print(b)
````

5. List: append() vs extend()

**Question:** How do `append()` and `extend()` behave differently?

**Python code:**

```python
a = [1, 2]
b = [1, 2]

a.append([3, 4])
b.extend([3, 4])

print(a)
print(b)
```

**Expected output:**

```text
[1, 2, [3, 4]]
[1, 2, 3, 4]
```

**Explanation:** `append()` adds its argument as one item, so the list becomes nested. `extend()` iterates over the supplied iterable and adds its elements individually.

````python
a = [1, 2]
b = [1, 2]

a.append([3, 4])
b.extend([3, 4])

print(a)
print(b)
````

6. List: Aliasing vs Copying

**Question:** What happens when two variables refer to the same list, compared with a sliced copy?

**Python code:**

```python
a = [1, 2, 3]
b = a
c = a[:]

a[0] = 100

print(a)
print(b)
print(c)
```

**Expected output:**

```text
[100, 2, 3]
[100, 2, 3]
[1, 2, 3]
```

**Explanation:** `b = a` does not create a new list; both names refer to the same object. `a[:]` creates a shallow copy of the outer list, so changing an element of `a` does not change `c` in this example.

````python
a = [1, 2, 3]
b = a
c = a[:]

a[0] = 100

print(a)
print(b)
print(c)
````

7. Set: Remove Duplicates and Sort

**Question:** How can you remove duplicate numbers and display the unique values in ascending order?

**Python code:**

```python
numbers = [1, 2, 2, 3, 3, 3, 4]

unique_numbers = sorted(set(numbers))
print(unique_numbers)
```

**Expected output:**

```text
[1, 2, 3, 4]
```

**Explanation:** A set keeps unique elements. `sorted()` returns those elements in ascending order as a list. Sets themselves should not be relied on for display order.

````python
numbers = [1, 2, 2, 3, 3, 3, 4]

unique_numbers = sorted(set(numbers))
print(unique_numbers)
````

8. Set: Union, Intersection, and Difference

**Question:** Find the elements in either set, in both sets, and only in the first set.

**Python code:**

```python
a = {1, 2, 3, 4}
b = {3, 4, 5, 6}

print(sorted(a | b))
print(sorted(a & b))
print(sorted(a - b))
```

**Expected output:**

```text
[1, 2, 3, 4, 5, 6]
[3, 4]
[1, 2]
```

**Explanation:** `|` gives the union, `&` gives the intersection, and `-` gives the difference (elements in `a` but not in `b`). Sorting makes the displayed output predictable.

````python
a = {1, 2, 3, 4}
b = {3, 4, 5, 6}

print(sorted(a | b))
print(sorted(a & b))
print(sorted(a - b))
````

9. String: Palindrome Check

**Question:** Check whether a string reads the same forward and backward, ignoring case and spaces.

**Python code:**

```python
text = "Never Odd Or Even"
cleaned = "".join(text.lower().split())

if cleaned == cleaned[::-1]:
    print("Palindrome")
else:
    print("Not a palindrome")
```

**Expected output:**

```text
Palindrome
```

**Explanation:** `lower()` makes the check case-insensitive. `split()` separates the text at whitespace, and `join()` combines the pieces without spaces. `[::-1]` reverses the resulting string.

````python
text = "Never Odd Or Even"
cleaned = "".join(text.lower().split())

if cleaned == cleaned[::-1]:
    print("Palindrome")
else:
    print("Not a palindrome")
````

10. Core Python: Mutable Default Argument Trap

**Question:** Why do repeated calls to this function share the same list, and how can you avoid it?

**Python code:**

```python
def add_item(item, items=None):
    if items is None:
        items = []
    items.append(item)
    return items

print(add_item("apple"))
print(add_item("banana"))
print(add_item("cherry"))
```

**Expected output:**

```text
['apple']
['banana']
['cherry']
```

**Explanation:** A mutable default such as `items=[]` is created once when the function is defined and reused across calls. Using `None` as the default and creating a new list inside the function avoids this shared-state bug.

````python
def add_item(item, items=None):
    if items is None:
        items = []
    items.append(item)
    return items

print(add_item("apple"))
print(add_item("banana"))
print(add_item("cherry"))
````
