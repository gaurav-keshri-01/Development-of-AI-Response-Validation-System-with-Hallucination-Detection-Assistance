```python
import sqlite3
import pandas as pd
from IPython.display import display

# Create an in-memory SQLite database.
# It exists for this notebook session and requires no separate installation.
conn = sqlite3.connect(":memory:")

# Create and populate the Employees table.
conn.executescript("""
DROP TABLE IF EXISTS Employees;
DROP TABLE IF EXISTS Departments;

CREATE TABLE Employees (
    id INTEGER PRIMARY KEY,
    name TEXT NOT NULL,
    department TEXT NOT NULL,
    salary INTEGER NOT NULL,
    city TEXT NOT NULL
);

INSERT INTO Employees (id, name, department, salary, city) VALUES
(1, 'Amit',  'IT',    60000, 'Pune'),
(2, 'Priya', 'HR',    40000, 'Mumbai'),
(3, 'Rahul', 'IT',    80000, 'Pune'),
(4, 'Neha',  'HR',    50000, 'Delhi'),
(5, 'Karan', 'Sales', 70000, 'Mumbai'),
(6, 'Sneha', 'IT',    60000, 'Delhi'),
(7, 'Vikas', 'Sales', 30000, 'Pune');

CREATE TABLE Departments (
    department TEXT PRIMARY KEY,
    location TEXT NOT NULL
);

INSERT INTO Departments (department, location) VALUES
('IT', 'Bangalore'),
('HR', 'Mumbai'),
('Sales', 'Delhi');
""")

print("Database ready. Employees and Departments tables have been created.")


```

```python
query = """SELECT name, salary
FROM Employees
WHERE salary > 50000
ORDER BY salary DESC, name;"""
result = pd.read_sql_query(query, conn)
display(result)
```

```python
query = """SELECT MAX(salary) AS second_highest_salary
FROM Employees
WHERE salary < (SELECT MAX(salary) FROM Employees);"""
result = pd.read_sql_query(query, conn)
display(result)
```

```python
query = """SELECT department, COUNT(*) AS total_employees
FROM Employees
GROUP BY department
ORDER BY department;"""
result = pd.read_sql_query(query, conn)
display(result)
```

```python
query = """SELECT department, ROUND(AVG(salary), 2) AS average_salary
FROM Employees
GROUP BY department
HAVING AVG(salary) > 50000
ORDER BY department;"""
result = pd.read_sql_query(query, conn)
display(result)
```

```python
query = """SELECT salary, COUNT(*) AS occurrences
FROM Employees
GROUP BY salary
HAVING COUNT(*) > 1
ORDER BY salary;"""
result = pd.read_sql_query(query, conn)
display(result)
```

```python
query = """SELECT name, department
FROM Employees
WHERE name LIKE 'A%'
ORDER BY name;"""
result = pd.read_sql_query(query, conn)
display(result)
```

```python
query = """SELECT name, salary
FROM Employees
WHERE salary > (SELECT AVG(salary) FROM Employees)
ORDER BY salary DESC, name;"""
result = pd.read_sql_query(query, conn)
display(result)
```

```python
query = """SELECT name, department, salary
FROM Employees AS e
WHERE salary = (
    SELECT MAX(salary)
    FROM Employees
    WHERE department = e.department
)
ORDER BY department, name;"""
result = pd.read_sql_query(query, conn)
display(result)
```

```python
query = """SELECT e.name, e.department, d.location
FROM Employees AS e
INNER JOIN Departments AS d
    ON e.department = d.department
ORDER BY e.id;"""
result = pd.read_sql_query(query, conn)
display(result)
```

```python
query = """SELECT
    name,
    salary,
    RANK() OVER (ORDER BY salary DESC) AS salary_rank
FROM Employees
ORDER BY salary DESC, name;"""
result = pd.read_sql_query(query, conn)
display(result)
```
